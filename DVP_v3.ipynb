```python
wb['Country Name'] = (
    wb['Country Name']
    .str.replace(r'\(.*\)', '', regex=True)
    .str.replace(',', '', regex=False)
    .str.strip()
)
```

```python
#removinf aggrgated regions and group (same as OWID cleaning process)
exclude_keywords_wb = [
    "income", "least", "world", "union", "gcp", "excl", "oecd", "members",
    "transport", "international", "region", "area", "dividend", "fragile",
    "asia", "africa", "europe", "america", "caribbean", "pacific",
    "south asia", "north america", "middle east", "central", "east", "west"
]

mask_agg = ~wb['Country Name'].str.lower().apply(
    lambda x: any(kw in x for kw in exclude_keywords_wb)
)
```

```python
#removing small territories - to keep only soverein nations
exclude_territories_wb = [
    "Aruba","Bermuda","British Virgin Islands","Curacao","Cayman Islands",
    "Channel Islands","Gibraltar","Greenland","Guam","Isle of Man",
    "New Caledonia","Northern Mariana Islands","Puerto Rico (US)",
    "Turks and Caicos Islands","French Polynesia","Virgin Islands (U.S.)",
    "American Samoa","St. Martin (French part)","Faroe Islands"
]

mask_terr = ~wb['Country Name'].isin(exclude_territories_wb)
```

```python
wb_clean = wb[mask_agg & mask_terr].copy()
```

```python
wb_clean['Country Name'].unique()
```

```python
drop_names = {
    "Heavily indebted poor countries", "IBRD only", "IDA only", "IDA blend",
    "IDA total", "IDA & IBRD total", "Not classified", "Other small states",
    "Small states", "Puerto Rico", "Virgin Islands", "Sint Maarten", "St. Martin"
}
wb_clean = wb_clean[~wb_clean["Country Name"].isin(drop_names)].copy()
```

```python
import pycountry
import pycountry_convert as pc

def country_to_continent(alpha3):
    try:
        country = pycountry.countries.get(alpha_3=alpha3)
        if not country:
            return None
        alpha2 = country.alpha_2
        continent_code = pc.country_alpha2_to_continent_code(alpha2)
        continent_dict = {
            "AF": "Africa",
            "AS": "Asia",
            "EU": "Europe",
            "NA": "North America",
            "SA": "South America",
            "OC": "Oceania"
        }
        return continent_dict.get(continent_code, None)
    except:
        return None

co2_annual_country["Continent"] = co2_annual_country["Code"].apply(country_to_continent)
```
